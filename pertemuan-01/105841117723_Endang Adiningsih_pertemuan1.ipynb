## Tugas 1 – Statistik Array Random

Pada bagian ini saya membuat array berisi 20 angka acak menggunakan NumPy.
Kemudian saya menghitung nilai mean, median, dan standar deviasi dari data tersebut.

```python
# Tugas 1: array random 20 angka, hitung mean, median, std
print("\n=== Tugas 1: Statistik Array Random ===")
np.random.seed(7)
arr = np.random.uniform(10, 90, 20)

print("Array:", np.round(arr, 2))
print("Mean  :", round(arr.mean(), 2))
print("Median:", round(np.median(arr), 2))
print("Std   :", round(arr.std(), 2))
```

## Tugas 2 – Data Nilai Mahasiswa

Pada bagian ini saya membuat DataFrame yang berisi 10 mahasiswa dengan nilai acak antara 60 sampai 95.
Data dibuat menggunakan Pandas agar tersusun dalam bentuk tabel.
Kemudian saya mencari 3 mahasiswa dengan nilai tertinggi menggunakan fungsi nlargest().

```python
# Tugas 2: DataFrame 10 mahasiswa, cari 3 nilai tertinggi
print("\n=== Tugas 2: Top 3 Nilai ===")
np.random.seed(21)

df_mhs = pd.DataFrame({
    "nama": [f"Mahasiswa_{i+1}" for i in range(10)],
    "nilai": np.random.randint(60, 95, 10)
})

print(df_mhs)

top3 = df_mhs.nlargest(3, "nilai")

print("\nTop 3 Nilai Tertinggi:")
print(top3)
```

## Tugas 3 – Visualisasi Data

Pada bagian ini saya membuat dua jenis grafik dari data nilai mahasiswa.
Grafik pertama adalah bar chart untuk melihat perbandingan nilai setiap mahasiswa.
Grafik kedua adalah histogram untuk melihat distribusi nilai secara keseluruhan.
Visualisasi ini membantu memahami pola data dengan lebih jelas.

```python
# ===============================
# Tugas 3: Membuat 2 Grafik Nilai Mahasiswa
# ===============================

import matplotlib.pyplot as plt

# Membuat figure dengan 2 subplot (1 baris, 2 kolom)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# -------------------------------
# Grafik 1: Bar Chart
# -------------------------------
axes[0].bar(df_mhs["nama"], df_mhs["nilai"], color="steelblue")
axes[0].set_title("Bar Chart Nilai Mahasiswa")
axes[0].set_xlabel("Nama Mahasiswa")
axes[0].set_ylabel("Nilai")
axes[0].tick_params(axis="x", rotation=45)

# -------------------------------
# Grafik 2: Histogram
# -------------------------------
axes[1].hist(df_mhs["nilai"], bins=5, color="coral", edgecolor="black")
axes[1].set_title("Histogram Distribusi Nilai")
axes[1].set_xlabel("Rentang Nilai")
axes[1].set_ylabel("Frekuensi")

plt.tight_layout()
plt.savefig("pertemuan01_tugas_grafik.png")
plt.show()

print("\n=== Tugas 3 ===")
print("Grafik berhasil disimpan sebagai 'pertemuan01_tugas_grafik.png'")
```

Penjelasan Grafik:
Bar chart ini menunjukkan perbandingan nilai dari 10 mahasiswa.
Histogram menunjukkan distribusi nilai sehingga terlihat sebaran nilai keseluruhan.
