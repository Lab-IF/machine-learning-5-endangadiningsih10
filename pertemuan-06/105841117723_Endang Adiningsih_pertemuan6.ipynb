# Pertemuan 06 – Ensemble Methods

## Tujuan
1. Membandingkan Decision Tree, Random Forest, dan Gradient Boosting.
2. Menampilkan 10 fitur terpenting dari Random Forest.
3. Melihat pengaruh perubahan n_estimators terhadap akurasi.

# 1️⃣ Perbandingan Model Klasifikasi

Pada tugas ini dibandingkan tiga model:
- Decision Tree (baseline)
- Random Forest (bagging)
- Gradient Boosting (boosting)

Dataset: Breast Cancer Wisconsin
Metric: Accuracy

```python
# ============================================================
# Tugas 1: Bandingkan DT vs RF vs GB (Classification)
# ============================================================
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score

# Load dataset
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

# Model
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}

print("="*50)
print("HASIL PERBANDINGAN MODEL")
print("="*50)

for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    acc = accuracy_score(y_test, pred)
    print(f"{name:<20} | Accuracy = {acc:.3f}")
```

# 2️⃣ Feature Importance (Random Forest)

Menampilkan 10 fitur paling penting berdasarkan nilai feature_importances_
dari model Random Forest.

```python
# ============================================================
# Tugas 2: Top 10 Feature Importance dari Random Forest
# ============================================================
import pandas as pd

# Latih Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# Ambil feature importance
importances = rf.feature_importances_
features = data.feature_names

# Buat DataFrame
feat_importance = pd.DataFrame({
    "Feature": features,
    "Importance": importances
})

# Urutkan dari terbesar
feat_importance = feat_importance.sort_values(by="Importance", ascending=False)

print("="*50)
print("10 FITUR TERPENTING (Random Forest)")
print("="*50)

print(feat_importance.head(10))
```

# 3️⃣ Pengaruh n_estimators

Menguji Random Forest dengan jumlah pohon berbeda:
- 10
- 50
- 100
- 200

Melihat pengaruhnya terhadap accuracy.

```python
# ============================================================
# Tugas 3: Pengaruh n_estimators Pada Random Forest
# ============================================================
print("="*55)
print("PENGARUH n_estimators TERHADAP AKURASI")
print("="*55)

results = {}

for n in [10, 50, 100, 200]:
    rf = RandomForestClassifier(n_estimators=n, random_state=42)
    rf.fit(X_train, y_train)
    pred = rf.predict(X_test)
    acc = accuracy_score(y_test, pred)
    results[n] = acc
    print(f"n_estimators = {n:<3} | Accuracy = {acc:.3f}")

print("\n" + "="*55)
print("ANALISIS SINGKAT")
print("="*55)

best_n = max(results, key=results.get)
best_acc = results[best_n]

print(f"Model terbaik diperoleh saat n_estimators = {best_n}")
print(f"Dengan accuracy sebesar {best_acc:.3f}")

print("\nPenjelasan:")
print("- Semakin banyak jumlah pohon, model menjadi lebih stabil.")
print("- Dari 50 ke atas, peningkatan akurasi tidak signifikan.")
print("- Artinya model sudah cukup optimal mulai n_estimators = 50.")
```

## Kesimpulan

1. Random Forest dan Gradient Boosting menghasilkan akurasi lebih tinggi
   dibandingkan Decision Tree tunggal karena teknik ensemble
   mampu mengurangi overfitting dan meningkatkan stabilitas model.

2. Random Forest menggunakan metode bagging (banyak pohon paralel),
   sedangkan Gradient Boosting menggunakan metode boosting
   (pohon dilatih berurutan untuk memperbaiki kesalahan sebelumnya).

3. Penambahan n_estimators cenderung meningkatkan stabilitas
   dan sedikit meningkatkan akurasi, namun waktu komputasi juga bertambah.
